```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
from statsmodels.nonparametric.smoothers_lowess import lowess

# Data Cleaning
MASTER_CLOSE_PRICE_DATA = MASTER_CLOSE_PRICE_DATA.drop_duplicates()
MASTER_CLOSE_PRICE_DATA.index = pd.to_datetime(MASTER_CLOSE_PRICE_DATA.index)
MASTER_CLOSE_PRICE_DATA = MASTER_CLOSE_PRICE_DATA.sort_index()

# Filling Missing Data
# When you use method='linear', it performs linear interpolation, which estimates the missing 
# values by connecting the data points with straight lines. This method avoids the creation of 
# repeated values that can occur with forward fill (ffill) and backward fill (bfill).
# Linear interpolation provides a smooth transition between known data points, 
# avoiding abrupt changes or repeated values.
MASTER_CLOSE_PRICE_DATA = MASTER_CLOSE_PRICE_DATA.interpolate(method='linear')

# Simple Statistical Analysis
MASTER_CLOSE_PRICE_DATA_STATS = MASTER_CLOSE_PRICE_DATA.describe()
MASTER_CLOSE_PRICE_DATA_RETURNS = MASTER_CLOSE_PRICE_DATA.pct_change()
MASTER_CLOSE_PRICE_DATA_RM = MASTER_CLOSE_PRICE_DATA.rolling(window=20).mean()
MASTER_CLOSE_PRICE_DATA_RS = MASTER_CLOSE_PRICE_DATA.rolling(window=20).std()

# Normalization Methods
MASTER_CLOSE_PRICE_DATA_NL = pd.DataFrame(index=MASTER_CLOSE_PRICE_DATA.index)
for column in MASTER_CLOSE_PRICE_DATA.columns:
    MASTER_CLOSE_PRICE_DATA_NL[f'{column}_Log'] = np.log(MASTER_CLOSE_PRICE_DATA[column])

MASTER_CLOSE_PRICE_DATA_NZS = pd.DataFrame(index=MASTER_CLOSE_PRICE_DATA.index)
for column in MASTER_CLOSE_PRICE_DATA.columns:
    MASTER_CLOSE_PRICE_DATA_NZS[f'{column}_Z_Score'] = stats.zscore(MASTER_CLOSE_PRICE_DATA[column])

# Smoothing Algorithms
MASTER_CLOSE_PRICE_DATA_SMOOTHED = pd.DataFrame(index=MASTER_CLOSE_PRICE_DATA.index)
for column in MASTER_CLOSE_PRICE_DATA.columns:
    MASTER_CLOSE_PRICE_DATA_SMOOTHED[f'{column}_SMA'] = MASTER_CLOSE_PRICE_DATA[column].rolling(window=20).mean()
    MASTER_CLOSE_PRICE_DATA_SMOOTHED[f'{column}_EMA'] = MASTER_CLOSE_PRICE_DATA[column].ewm(alpha=0.1, adjust=False).mean()
    MASTER_CLOSE_PRICE_DATA_SMOOTHED_LOWESS = lowess(MASTER_CLOSE_PRICE_DATA[column], np.arange(len(MASTER_CLOSE_PRICE_DATA)), frac=0.1)
    MASTER_CLOSE_PRICE_DATA_SMOOTHED[f'{column}_LOWESS'] = MASTER_CLOSE_PRICE_DATA_SMOOTHED_LOWESS[:, 1]
```

```python
from statsmodels.tsa.stattools import acf, pacf

# Calculate autocorrelation and confidence intervals
lags = 20  # Number of lags to calculate
autocorr_values, confint = acf(MASTER_CLOSE_PRICE_DATA['AAPL'], nlags=lags, alpha=0.05)

# Extract the lower and upper bounds of the confidence intervals
lower_bound = confint[:, 0] - autocorr_values
upper_bound = confint[:, 1] - autocorr_values

# Plot the time series
fig = go.Figure()

fig.add_trace(go.Scatter(x=MASTER_CLOSE_PRICE_DATA.index, y=MASTER_CLOSE_PRICE_DATA['AAPL'], mode='lines', name='Close Prices'))

# Add title and labels
fig.update_layout(
  title='Time Series Data (AAPL)',
  xaxis_title='Date',
  yaxis_title='Close Price',
  template='plotly_dark'
)

# Show the plot
fig.show()

# Plot autocorrelation with confidence intervals
fig_acf = go.Figure()

fig_acf.add_trace(go.Bar(x=list(range(lags + 1)), y=autocorr_values, name='Autocorrelation'))

# Add confidence intervals
fig_acf.add_trace(go.Scatter(
  x=list(range(lags + 1)),
  y=upper_bound,
  mode='lines',
  line=dict(color='red', dash='dash'),
  name='Upper Bound'
))

fig_acf.add_trace(go.Scatter(
  x=list(range(lags + 1)),
  y=lower_bound,
  mode='lines',
  line=dict(color='orange', dash='dash'),
  name='Lower Bound'
))

# Add title and labels
fig_acf.update_layout(
  title='Autocorrelation of Close Prices with Confidence Intervals (AAPL)',
  xaxis_title='Lag',
  yaxis_title='Autocorrelation',
  template='plotly_dark'
)

# Show the plot
fig_acf.show()

# Calculate partial autocorrelation
pacf_values, pacf_confint = pacf(MASTER_CLOSE_PRICE_DATA['AAPL'], nlags=lags, alpha=0.05)

# Extract the lower and upper bounds of the confidence intervals for PACF
pacf_lower_bound = pacf_confint[:, 0] - pacf_values
pacf_upper_bound = pacf_confint[:, 1] - pacf_values

# Plot partial autocorrelation with confidence intervals
fig_pacf = go.Figure()

fig_pacf.add_trace(go.Bar(x=list(range(lags + 1)), y=pacf_values, name='Partial Autocorrelation'))

# Add confidence intervals for PACF
fig_pacf.add_trace(go.Scatter(
  x=list(range(lags + 1)),
  y=pacf_upper_bound,
  mode='lines',
  line=dict(color='red', dash='dash'),
  name='Upper Bound'
))

fig_pacf.add_trace(go.Scatter(
  x=list(range(lags + 1)),
  y=pacf_lower_bound,
  mode='lines',
  line=dict(color='orange', dash='dash'),
  name='Lower Bound'
))

# Add title and labels
fig_pacf.update_layout(
  title='Partial Autocorrelation of Close Prices with Confidence Intervals (AAPL)',
  xaxis_title='Lag',
  yaxis_title='Partial Autocorrelation',
  template='plotly_dark'
)

# Show the plot
fig_pacf.show()
```

```python
import pandas as pd
import numpy as np
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import plotly.graph_objects as go

# Extracting only the GSK column.
GSK_Master_Data = MASTER_CLOSE_PRICE_DATA['GSK']

# Ensuring the index is a DatetimeIndex
if not isinstance(GSK_Master_Data.index, pd.DatetimeIndex):
    GSK_Master_Data.index = pd.to_datetime(GSK_Master_Data.index)

# Calculating the split point
split_point = int(len(GSK_Master_Data) * 0.8)

# Splitting the data into Train and Test
train_GSK_data = GSK_Master_Data[:split_point]
test_GSK_data = GSK_Master_Data[split_point:]

# Exponential Weighted Moving Average model
alpha = 0.3  # we can adjust this smoothing factor alpha which means tweaking the model
train_GSK_data_ewma = train_GSK_data.ewm(alpha=alpha).mean()

# Making predictions for the test period and future 180 days additionally
forecast_GSK = []
last_value = train_GSK_data_ewma.iloc[-1]
for _ in range(len(test_GSK_data)):
    forecast_GSK.append(last_value)
    last_value = alpha * test_GSK_data.iloc[_] + (1 - alpha) * last_value

# Calculating the average daily change in the last 30 days of predictions
avg_daily_change_GSK = np.mean(np.diff(forecast_GSK[-30:]))

# Adding future predictions with fluctuations
np.random.seed(42)  # for reproducibility
std_dev_GSK = train_GSK_data.std()
future_days = 180
for _ in range(future_days):
    # Add the average daily change plus some random noise
    next_value = last_value + avg_daily_change_GSK + np.random.normal(0, std_dev_GSK/2)
    forecast_GSK.append(next_value)
    last_value = next_value

# Creating date range for future predictions
future_dates_GSK = pd.date_range(start=test_GSK_data.index[-1] + pd.Timedelta(days=1), periods=future_days)
forecast_index = test_GSK_data.index.union(future_dates_GSK)

forecast_df_GSK = pd.DataFrame(forecast_GSK, index=forecast_index, columns=['Forecast'])

# Calculating confidence intervals (using standard deviation of training data)
std_dev_GSK = train_GSK_data.std()
lower_ci_GSK = forecast_df_GSK['Forecast'] - 1.96 * std_dev_GSK
upper_ci_GSK = forecast_df_GSK['Forecast'] + 1.96 * std_dev_GSK

# Function to calculate MAPE
def mean_absolute_percentage_error(y_true, y_pred):
    return np.mean(np.abs((y_true - y_pred) / y_true)) * 100

# Function to calculate RMSLE
def root_mean_squared_log_error(y_true, y_pred):
    return np.sqrt(np.mean(np.square(np.log1p(y_true) - np.log1p(y_pred))))

# Evaluating the model for training period
mse_train_GSK = mean_squared_error(train_GSK_data, train_GSK_data_ewma)
rmse_train_GSK = np.sqrt(mse_train_GSK)
mae_train_GSK = mean_absolute_error(train_GSK_data, train_GSK_data_ewma)
r2_train_GSK = r2_score(train_GSK_data, train_GSK_data_ewma)

# Evaluating the model for testing period
mse_test_GSK = mean_squared_error(test_GSK_data, forecast_df_GSK['Forecast'][:len(test_GSK_data)])
rmse_test_GSK = np.sqrt(mse_test_GSK)
mae_test_GSK = mean_absolute_error(test_GSK_data, forecast_df_GSK['Forecast'][:len(test_GSK_data)])
r2_test_GSK = r2_score(test_GSK_data, forecast_df_GSK['Forecast'][:len(test_GSK_data)])

print("Training Metrics:")
print(f"MSE for GSK using EWMA: {mse_train_GSK}")
print(f"RMSE for GSK using EWMA: {rmse_train_GSK}")
print(f"MAE for GSK using EWMA: {mae_train_GSK}")
print(f"R2 for GSK using EWMA: {r2_train_GSK}")

print("\nTesting Metrics:")
print(f"MSE for GSK using EWMA: {mse_test_GSK}")
print(f"RMSE for GSK using EWMA: {rmse_test_GSK}")
print(f"MAE for GSK using EWMA: {mae_test_GSK}")
print(f"R2 for GSK using EWMA: {r2_test_GSK}")

# Creating the plot: Actual vs Predicted for GSK with confidence intervals and future predictions
fig1 = go.Figure()

fig1.add_trace(go.Scatter(x=GSK_Master_Data.index, y=GSK_Master_Data, mode='lines', name='Actual', line=dict(color='blue')))
fig1.add_trace(go.Scatter(x=train_GSK_data.index, y=train_GSK_data, mode='lines', name='Train', line=dict(color='green')))
fig1.add_trace(go.Scatter(x=test_GSK_data.index, y=test_GSK_data, mode='lines', name='Test', line=dict(color='red')))
fig1.add_trace(go.Scatter(x=forecast_df_GSK.index, y=forecast_df_GSK['Forecast'], mode='lines', name='Predictions', line=dict(color='orange')))

# Adding confidence intervals
fig1.add_trace(go.Scatter(
    x=forecast_df_GSK.index.tolist()+forecast_df_GSK.index.tolist()[::-1],
    y=upper_ci_GSK.tolist()+lower_ci_GSK.tolist()[::-1],
    fill='toself',
    fillcolor='rgba(0,100,80,0.2)',
    line=dict(color='rgba(255,255,255,0)'),
    hoverinfo="skip",
    showlegend=False
))

# Adding vertical line to separate actual data from future predictions
fig1.add_vline(x=test_GSK_data.index[-1], line_dash="dash", line_color="gray")

fig1.update_layout(
    height=600, 
    title_text="GSK Stock Price - Actual vs Predicted (EWMA Model) with Confidence Intervals and Future Predictions",
    xaxis_title="Date",
    yaxis_title="Stock Price",
    legend_title="Legend"
)
fig1.update_xaxes(type='date')
fig1.show()

# Creating the residuals plot (only for the test period)
fig2 = go.Figure()

residuals_GSK = test_GSK_data - forecast_df_GSK['Forecast'][:len(test_GSK_data)]
fig2.add_trace(go.Scatter(x=test_GSK_data.index, y=residuals_GSK, mode='lines', name='Residuals', line=dict(color='purple')))
fig2.add_shape(type="line", x0=test_GSK_data.index[0], y0=0, x1=test_GSK_data.index[-1], y1=0,
               line=dict(color="red", width=2, dash="dash"))

fig2.update_layout(
    height=400, 
    title_text="Residuals (EWMA Model)",
    xaxis_title="Date",
    yaxis_title="Residuals",
    showlegend=False
)
fig2.update_xaxes(type='date')
fig2.show()

# Printing some diagnostic information
print(f"First few values of train_data: {train_GSK_data.head()}")
print(f"First few values of test_data: {test_GSK_data.head()}")
print(f"First few values of forecast: {forecast_df_GSK.head()}")
print(f"Last few values of forecast (including future predictions): {forecast_df_GSK.tail()}")
print(f"First few values of residuals: {residuals_GSK.head()}")
```

```python
from scipy.optimize import minimize

# Extract only the CAH column
data = MASTER_CLOSE_PRICE_DATA['AAPL'].copy()

# Ensure the index is a DatetimeIndex
if not isinstance(data.index, pd.DatetimeIndex):
    data.index = pd.to_datetime(data.index)

# Calculate returns
returns = 100 * data.pct_change().dropna()

# Print data information
print(f"Data shape: {returns.shape}")
print(f"Date range: {returns.index.min()} to {returns.index.max()}")

# Split the data
split_point = int(len(returns) * 0.8)
train_returns = returns[:split_point]
test_returns = returns[split_point:]

# GARCH(1,1) log-likelihood function
def garch_likelihood(params, returns):
    omega, alpha, beta = params
    n = len(returns)
    sigma2 = np.zeros(n)
    sigma2[0] = np.var(returns)
    for t in range(1, n):
        sigma2[t] = omega + alpha * returns[t-1]**2 + beta * sigma2[t-1]
    return np.sum(np.log(sigma2) + returns**2 / sigma2)

# Fit GARCH(1,1) model
def fit_garch(returns):
    initial_params = [np.var(returns), 0.1, 0.8]
    bounds = [(0, None), (0, 1), (0, 1)]
    result = minimize(garch_likelihood, initial_params, args=(returns,), bounds=bounds, method='L-BFGS-B')
    return result.x

# Calculate GARCH(1,1) volatility
def calculate_garch_vol(params, returns):
    omega, alpha, beta = params
    n = len(returns)
    sigma2 = np.zeros(n)
    sigma2[0] = np.var(returns)
    for t in range(1, n):
        sigma2[t] = omega + alpha * returns[t-1]**2 + beta * sigma2[t-1]
    return np.sqrt(sigma2)

# Forecast GARCH(1,1) volatility with noise
def forecast_garch_with_noise(params, returns, n_steps, noise_level=0.1):
    omega, alpha, beta = params
    last_var = omega + alpha * returns[-1]**2 + beta * np.var(returns)
    forecasts = np.zeros(n_steps)
    for i in range(n_steps):
        noise = np.random.normal(0, noise_level * np.sqrt(last_var))
        forecasts[i] = np.sqrt(last_var) + noise
        last_var = omega + alpha * (forecasts[i]**2) + beta * last_var
    return forecasts

# Fit the GARCH(1,1) model
params = fit_garch(train_returns)
print("Estimated GARCH parameters:")
print(f"omega: {params[0]:.6f}")
print(f"alpha: {params[1]:.6f}")
print(f"beta: {params[2]:.6f}")

# Calculate in-sample volatility
in_sample_vol = calculate_garch_vol(params, train_returns)

# Make predictions with noise
noise_level = 0.1  # Adjust this value to control the amount of noise
conditional_vol = forecast_garch_with_noise(params, train_returns, len(test_returns), noise_level)

# Evaluate the model
mse = mean_squared_error(test_returns**2, conditional_vol**2)
rmse = np.sqrt(mse)
mae = mean_absolute_error(test_returns.abs(), conditional_vol)

print(f"Mean Squared Error: {mse}")
print(f"Root Mean Squared Error: {rmse}")
print(f"Mean Absolute Error: {mae}")

# Calculate confidence intervals (assuming normal distribution)
conf_level = 0.95
z_score = np.abs(np.random.normal(0, 1, 10000)).mean()  # Approximation of 95% confidence interval
lower_ci = conditional_vol - z_score * conditional_vol / np.sqrt(2)
upper_ci = conditional_vol + z_score * conditional_vol / np.sqrt(2)

# Plot the results
fig = go.Figure()
fig.add_trace(go.Scatter(x=returns.index, y=returns, mode='lines', name='Actual Returns', line=dict(color='blue')))
fig.add_trace(go.Scatter(x=train_returns.index, y=in_sample_vol, mode='lines', name='In-sample Volatility', line=dict(color='green')))
fig.add_trace(go.Scatter(x=test_returns.index, y=conditional_vol, mode='lines', name='Predicted Volatility', line=dict(color='red')))
fig.add_trace(go.Scatter(x=test_returns.index, y=upper_ci, mode='lines', name='Upper CI', line=dict(color='rgba(255,0,0,0.2)')))
fig.add_trace(go.Scatter(x=test_returns.index, y=lower_ci, mode='lines', name='Lower CI', line=dict(color='rgba(255,0,0,0.2)'), fill='tonexty'))
fig.update_layout(
    title='GARCH(1,1) Volatility Forecast (AAPL)',
    xaxis_title='Date',
    yaxis_title='Returns / Volatility',
    legend_title='Legend'
)
fig.show()

# Plot the volatility forecast
fig_forecast = go.Figure()
fig_forecast.add_trace(go.Scatter(x=test_returns.index, y=test_returns.abs(), mode='lines', name='Absolute Returns', line=dict(color='blue')))
fig_forecast.add_trace(go.Scatter(x=test_returns.index, y=conditional_vol, mode='lines', name='Predicted Volatility', line=dict(color='red')))
fig_forecast.add_trace(go.Scatter(x=test_returns.index, y=upper_ci, mode='lines', name='Upper CI', line=dict(color='rgba(255,0,0,0.2)')))
fig_forecast.add_trace(go.Scatter(x=test_returns.index, y=lower_ci, mode='lines', name='Lower CI', line=dict(color='rgba(255,0,0,0.2)'), fill='tonexty'))
fig_forecast.update_layout(
    title='GARCH(1,1) Volatility Forecast vs Absolute Returns (AAPL)',
    xaxis_title='Date',
    yaxis_title='Absolute Returns / Volatility',
    legend_title='Legend'
)
fig_forecast.show()

# Print some diagnostic information
print(f"\nFirst few values of train_returns:\n{train_returns.head()}")
print(f"\nFirst few values of test_returns:\n{test_returns.head()}")
print(f"\nFirst few values of predicted volatility:\n{conditional_vol[:5]}")
print(f"\nLast few values of predicted volatility:\n{conditional_vol[-5:]}")

# Forecast future volatility with noise
future_steps = 30  # Forecast for the next 30 days
future_volatility = forecast_garch_with_noise(params, returns, future_steps, noise_level)

# Calculate confidence intervals for future forecast
future_lower_ci = future_volatility - z_score * future_volatility / np.sqrt(2)
future_upper_ci = future_volatility + z_score * future_volatility / np.sqrt(2)

# Plot future volatility forecast
fig_future = go.Figure()
fig_future.add_trace(go.Scatter(x=pd.date_range(start=returns.index[-1], periods=future_steps+1, freq='D')[1:], 
                                y=future_volatility, mode='lines', name='Future Volatility Forecast', line=dict(color='green')))
fig_future.add_trace(go.Scatter(x=pd.date_range(start=returns.index[-1], periods=future_steps+1, freq='D')[1:], 
                                y=future_upper_ci, mode='lines', name='Upper CI', line=dict(color='rgba(0,255,0,0.2)')))
fig_future.add_trace(go.Scatter(x=pd.date_range(start=returns.index[-1], periods=future_steps+1, freq='D')[1:], 
                                y=future_lower_ci, mode='lines', name='Lower CI', line=dict(color='rgba(0,255,0,0.2)'), fill='tonexty'))
fig_future.update_layout(
    title='GARCH(1,1) Future Volatility Forecast (AAPL)',
    xaxis_title='Date',
    yaxis_title='Predicted Volatility',
    legend_title='Legend'
)
fig_future.show()

print(f"\nFuture volatility forecast for the next {future_steps} days:\n{future_volatility}")
```

## Third Model - Kalman Filter Model

```python
class KalmanFilter:
    def __init__(self, initial_state, initial_estimate_error, process_variance, measurement_variance):
        self.x = initial_state
        self.P = initial_estimate_error
        self.Q = process_variance
        self.R = measurement_variance

    def predict(self):
        self.x = self.x
        self.P = self.P + self.Q
        return self.x

    def update(self, measurement):
        self.K = self.P / (self.P + self.R)
        self.x = self.x + self.K * (measurement - self.x)
        self.P = (1 - self.K) * self.P

def apply_kalman_filter(data, process_variance, measurement_variance):
    kf = KalmanFilter(data[0], 1, process_variance, measurement_variance)
    filtered_data = []
    for measurement in data:
        kf.predict()
        kf.update(measurement)
        filtered_data.append(kf.x)
    return filtered_data
```

```python
# Extracting only the AAPL column.
AAPL_Master_Data = MASTER_CLOSE_PRICE_DATA_NZS['AAPL_Z_Score']

if not isinstance(AAPL_Master_Data.index, pd.DatetimeIndex):
    AAPL_Master_Data.index = pd.to_datetime(AAPL_Master_Data.index)

split_point = int(len(AAPL_Master_Data) * 0.8)
train_AAPL_data = AAPL_Master_Data[:split_point]
test_AAPL_data = AAPL_Master_Data[split_point:]

alpha = 0.3
train_AAPL_data_ewma = train_AAPL_data.ewm(alpha=alpha).mean()

forecast_AAPL = []
last_value = train_AAPL_data_ewma.iloc[-1]
for _ in range(len(test_AAPL_data)):
    forecast_AAPL.append(last_value)
    last_value = alpha * test_AAPL_data.iloc[_] + (1 - alpha) * last_value

avg_daily_change_AAPL = np.mean(np.diff(forecast_AAPL[-30:]))
std_dev_AAPL = train_AAPL_data.std()

np.random.seed(42)
future_days = 180
for _ in range(future_days):
    next_value = last_value + avg_daily_change_AAPL + np.random.normal(0, std_dev_AAPL/2)
    forecast_AAPL.append(next_value)
    last_value = next_value

future_dates_AAPL = pd.date_range(start=test_AAPL_data.index[-1] + pd.Timedelta(days=1), periods=future_days)
forecast_index = test_AAPL_data.index.union(future_dates_AAPL)
forecast_df_AAPL = pd.DataFrame(forecast_AAPL, index=forecast_index, columns=['Forecast'])

lower_ci_AAPL = forecast_df_AAPL['Forecast'] - 1.96 * std_dev_AAPL
upper_ci_AAPL = forecast_df_AAPL['Forecast'] + 1.96 * std_dev_AAPL

# New function to calculate MAPE
def mean_absolute_percentage_error(y_true, y_pred):
    return np.mean(np.abs((y_true - y_pred) / y_true)) * 100

# New function to calculate RMSLE
def root_mean_squared_log_error(y_true, y_pred):
    return np.sqrt(np.mean(np.square(np.log1p(y_true) - np.log1p(y_pred))))

# Evaluating the model for training period
mse_train_AAPL_1 = mean_squared_error(train_AAPL_data, train_AAPL_data_ewma)
rmse_train_AAPL_1 = np.sqrt(mse_train_AAPL_1)
mae_train_AAPL_1 = mean_absolute_error(train_AAPL_data, train_AAPL_data_ewma)
r2_train_AAPL_1 = r2_score(train_AAPL_data, train_AAPL_data_ewma)

# Evaluating the model for testing period
mse_test_AAPL_1 = mean_squared_error(test_AAPL_data, forecast_df_AAPL['Forecast'][:len(test_AAPL_data)])
rmse_test_AAPL_1 = np.sqrt(mse_test_AAPL_1)
mae_test_AAPL_1 = mean_absolute_error(test_AAPL_data, forecast_df_AAPL['Forecast'][:len(test_AAPL_data)])
r2_test_AAPL_1 = r2_score(test_AAPL_data, forecast_df_AAPL['Forecast'][:len(test_AAPL_data)])

print("Training Metrics:")
print(f"MSE for AAPL using EWMA: {mse_train_AAPL_1}")
print(f"RMSE for AAPL using EWMA: {rmse_train_AAPL_1}")
print(f"MAE for AAPL using EWMA: {mae_train_AAPL_1}")
print(f"R2 for AAPL using EWMA: {r2_train_AAPL_1}")

print("\nTesting Metrics:")
print(f"MSE for AAPL using EWMA: {mse_test_AAPL_1}")
print(f"RMSE for AAPL using EWMA: {rmse_test_AAPL_1}")
print(f"MAE for AAPL using EWMA: {mae_test_AAPL_1}")
print(f"R2 for AAPL using EWMA: {r2_test_AAPL_1}")

# Creating the plot: Actual vs Predicted for AAPL with confidence intervals and future predictions
fig1 = go.Figure()

fig1.add_trace(go.Scatter(x=AAPL_Master_Data.index, y=AAPL_Master_Data, mode='lines', 
                          name='Actual', line=dict(color='blue')))
fig1.add_trace(go.Scatter(x=train_AAPL_data.index, y=train_AAPL_data, mode='lines', 
                          name='Train', line=dict(color='green')))
fig1.add_trace(go.Scatter(x=test_AAPL_data.index, y=test_AAPL_data, mode='lines', 
                          name='Test', line=dict(color='red')))
fig1.add_trace(go.Scatter(x=forecast_df_AAPL.index, y=forecast_df_AAPL['Forecast'], 
                          mode='lines', name='Predictions', line=dict(color='orange')))

fig1.add_trace(go.Scatter(
    x=forecast_df_AAPL.index.tolist()+forecast_df_AAPL.index.tolist()[::-1],
    y=upper_ci_AAPL.tolist()+lower_ci_AAPL.tolist()[::-1],
    fill='toself',
    fillcolor='rgba(0,100,80,0.2)',
    line=dict(color='rgba(255,255,255,0)'),
    hoverinfo="skip",
    showlegend=False
))

fig1.add_vline(x=test_AAPL_data.index[-1], line_dash="dash", line_color="gray")

fig1.update_layout(
    height=600, 
    title_text="AAPL Stock Price - Actual vs Predicted (EWMA Model) with Confidence Intervals and Future Predictions",
    xaxis_title="Date",
    yaxis_title="Stock Price",
    legend_title="Legend"
)
fig1.update_xaxes(type='date')
fig1.show()

# Creating the residuals plot (only for the test period)
fig2 = go.Figure()

residuals_AAPL = test_AAPL_data - forecast_df_AAPL['Forecast'][:len(test_AAPL_data)]
fig2.add_trace(go.Scatter(x=test_AAPL_data.index, y=residuals_AAPL, mode='lines', name='Residuals', 
                          line=dict(color='purple')))
fig2.add_shape(type="line", x0=test_AAPL_data.index[0], y0=0, x1=test_AAPL_data.index[-1], y1=0,
               line=dict(color="red", width=2, dash="dash"))

fig2.update_layout(
    height=400, 
    title_text="Residuals (EWMA Model)",
    xaxis_title="Date",
    yaxis_title="Residuals",
    showlegend=False
)
fig2.update_xaxes(type='date')
fig2.show()

# Printing some diagnostic information
print(f"First few values of train_data: {train_AAPL_data.head()}")
print(f"First few values of test_data: {test_AAPL_data.head()}")
print(f"First few values of forecast: {forecast_df_AAPL.head()}")
print(f"Last few values of forecast (including future predictions): {forecast_df_AAPL.tail()}")
print(f"First few values of residuals: {residuals_AAPL.head()}")
```
